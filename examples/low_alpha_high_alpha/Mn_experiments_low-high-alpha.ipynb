```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib import rcParams
from chemevo import evolution as evo

params = {
   'axes.labelsize': 15,
   'font.size': 15,
   'legend.fontsize': 14,
   'xtick.minor.visible': True,
   'ytick.minor.visible': True,
   'xtick.labelsize': 15,
   'ytick.labelsize': 15,
   'text.usetex': True, #to use TeX in your labels
   'font.family':'serif',
   'axes.titlesize': 15,
   'xtick.direction': 'in',
   'ytick.direction': 'in',
   'xtick.top': True,
   'ytick.right': True,
   'xtick.major.size': 6,
   'xtick.minor.size': 3,
   'ytick.major.size': 6,
   'ytick.minor.size': 3
   }
rcParams.update(params)

colors = ["#0072B2", "#E69F00", "#009E73", "#D55E00", "#CC79A7", "#F0E442"]
```

```python
t_array = np.linspace(0,12,int(1e3))
tau_sfh = 6 #Gyrs
m_g_array_exp = 1 * np.exp(-t_array/tau_sfh)

#Gal1, with everything
gal1 = evo.Galaxy(t_array=t_array, m_g_array=m_g_array_exp)
Z_Fe_1 = gal1.compute_z_Fe()
Z_O_1 = gal1.compute_z_O()
Z_Mn_1 = gal1.compute_z_Mn()
eta_def = 2.5
tau_star_def = 1
#For high alpha, lower eta and higher tau_star
#For low alpha, the opposite
tau_star_for_high = 5
tau_star_for_low = 2
eta_for_high = 1
eta_for_low = 3

tau_sfh_def=6
tau_sfh_here=10

gal_high = evo.Galaxy(t_array=t_array, m_g_array=m_g_array_exp, eta=eta_for_high, tau_star=tau_star_for_high, tau_sfh=tau_sfh_here)
gal_low = evo.Galaxy(t_array=t_array, m_g_array=m_g_array_exp, eta=eta_for_low, tau_star=tau_star_for_low, tau_sfh=tau_sfh_here)
```

```python
plt.plot(gal1.O_H[1:], gal1.Mn_O[1:], label = r'$\eta = $' + f" {eta_def}, " + r'$\tau_{*} = $' + f"{tau_star_def}",
         color = 'gray', alpha = 0.5, linestyle='--')
plt.plot(gal_high.O_H[1:], gal_high.Mn_O[1:], label = r'$\eta = $' + f" {eta_for_high}, " + r'$\tau_{*} = $' + f"{tau_star_for_high}",
         color = 'red')
plt.plot(gal_low.O_H[1:], gal_low.Mn_O[1:], label = r'$\eta = $' + f" {eta_for_low}, " + r'$\tau_{*} = $' + f"{tau_star_for_low}", 
         color = 'blue')
plt.legend()
plt.xlabel('[O/H]')
plt.ylabel('[Mn/O]')
```

```python
plt.plot(gal1.O_H[1:], gal1.Fe_O[1:], label = r'$\eta = $' + f" {eta_def}, " + r'$\tau_{*} = $' + f"{tau_star_def}",
         color = 'gray', alpha = 0.5, linestyle='--')
plt.plot(gal_high.O_H[1:], gal_high.Fe_O[1:], label = r'$\eta = $' + f" {eta_for_high}, " + r'$\tau_{*} = $' + f"{tau_star_for_high}",
         color = 'red')
plt.plot(gal_low.O_H[1:], gal_low.Fe_O[1:], label = r'$\eta = $' + f" {eta_for_low}, " + r'$\tau_{*} = $' + f"{tau_star_for_low}", 
         color = 'blue')
plt.legend()
plt.xlabel('[O/H]')
plt.ylabel('[Fe/O]')
```
