```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib import rcParams
from chemevo import evolution as evo

params = {
   'axes.labelsize': 15,
   'font.size': 15,
   'legend.fontsize': 14,
   'xtick.minor.visible': True,
   'ytick.minor.visible': True,
   'xtick.labelsize': 15,
   'ytick.labelsize': 15,
   'text.usetex': True, #to use TeX in your labels
   'font.family':'serif',
   'axes.titlesize': 15,
   'xtick.direction': 'in',
   'ytick.direction': 'in',
   'xtick.top': True,
   'ytick.right': True,
   'xtick.major.size': 6,
   'xtick.minor.size': 3,
   'ytick.major.size': 6,
   'ytick.minor.size': 3
   }
rcParams.update(params)

colors = ["#0072B2", "#E69F00", "#009E73", "#D55E00", "#CC79A7", "#F0E442"]
linestyles = ['-', '--', ':', '-.']
```

```python
t_array = np.linspace(0.0001,12,int(1e3))
tau_sfh = 6 #Gyrs
m_g_array_exp = 1 * np.exp(-t_array/tau_sfh)
m_g_array_const = np.ones(len(t_array))
```

# 1) Star Formation Efficiency ($\eta$)

## 1.1) Constant SFR

```python
#gals c are constant SFR galaxies
gals_c = []
etas = [0.5, 1.5, 2.5, 5]

for eta in etas:
    gals_c.append(evo.Galaxy(t_array=t_array, m_g_array=m_g_array_const, eta=eta))

#gals e are exp SFR galaxies
gals_e = []

for eta in etas:
    gals_e.append(evo.Galaxy(t_array=t_array, m_g_array=m_g_array_exp, eta=eta))
```

```python
plt.figure(figsize=(12,4))
plt.subplot(1,3,1)
for i in range(len(etas)):
    plt.plot(t_array, gals_c[i].compute_z_Mn()/gals_c[i].SolarMn, label = r"$\eta = $" + f"{etas[i]}", color = colors[i])
plt.xlabel('Time (Gyrs)')
plt.ylabel(r'$Z_{Mn}/Z_{Mn, \odot}$')
plt.axhline(1, color = 'gray', linestyle='--')
plt.legend()

plt.subplot(1,3,2)
for i in range(len(etas)):
    plt.plot(t_array, gals_c[i].compute_z_Fe()/gals_c[i].SolarFe, label = r"$\eta = $" + f"{etas[i]}", color = colors[i])
plt.axhline(1, color = 'gray', linestyle='--')
plt.xlabel('Time (Gyrs)')
plt.ylabel(r'$Z_{Fe}/Z_{Fe, \odot}$')

plt.subplot(1,3,3)
for i in range(len(etas)):
    plt.plot(t_array, gals_c[i].compute_z_O()/gals_c[i].SolarO, label = r"$\eta = $" + f"{etas[i]}", color = colors[i])
plt.axhline(1, color = 'gray', linestyle='--')
plt.xlabel('Time (Gyrs)')
plt.ylabel(r'$Z_{O}/Z_{O, \odot}$')

plt.suptitle('Const SFR')
plt.tight_layout()
```

```python
plt.figure(figsize=(9,4))
plt.subplot(1,2,1)
for i in range(len(etas)):
    plt.plot(gals_c[i].O_H[1:], gals_c[i].Mn_O[1:], label = r"$\eta = $" + f"{etas[i]}", color = colors[i])
plt.xlabel('[O/H]')
plt.ylabel('[Mn/O]')
plt.legend()

plt.subplot(1,2,2)
for i in range(len(etas)):
    plt.plot(gals_c[i].O_H[1:], gals_c[i].Fe_O[1:], label = r"$\eta = $" + f"{etas[i]}", color = colors[i])
plt.xlabel('[O/H]')
plt.ylabel('[Fe/O]')

plt.suptitle('Const SFR')
plt.tight_layout()
```

## 1.2) Exp SFR

```python
plt.figure(figsize=(14,5))
plt.subplot(1,3,1)
for i in range(len(etas)):
    plt.plot(t_array, gals_e[i].compute_z_Mn()/gals_e[i].SolarMn, label = r"$\eta = $" + f"{etas[i]}", color = colors[i])
plt.xlabel('Time (Gyrs)')
plt.ylabel(r'$Z_{Mn}/Z_{Mn, \odot}$')
plt.axhline(1, color = 'gray', linestyle='--')
plt.legend()

plt.subplot(1,3,2)
for i in range(len(etas)):
    plt.plot(t_array, gals_e[i].compute_z_Fe()/gals_e[i].SolarFe, label = r"$\eta = $" + f"{etas[i]}", color = colors[i])
plt.axhline(1, color = 'gray', linestyle='--')
plt.xlabel('Time (Gyrs)')
plt.ylabel(r'$Z_{Fe}/Z_{Fe, \odot}$')

plt.subplot(1,3,3)
for i in range(len(etas)):
    plt.plot(t_array, gals_e[i].compute_z_O()/gals_e[i].SolarO, label = r"$\eta = $" + f"{etas[i]}", color = colors[i])
plt.axhline(1, color = 'gray', linestyle='--')
plt.xlabel('Time (Gyrs)')
plt.ylabel(r'$Z_{O}/Z_{O, \odot}$')

plt.suptitle('Exp SFR')
plt.tight_layout()
```
